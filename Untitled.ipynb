```python
import json
import h5py
import matplotlib.pyplot as plt

# 1. Read the JSON metadata
with open("20260405_224312_meta.json", "r") as f:
    meta = json.load(f)
    print(f"Controller used: {meta['controller']}")

# 2. Read the H5 physics data
with h5py.File("20260405_224312.h5", "r") as data:
    # Extract the arrays into Python variables
    t = data["t"][:]       # Time
    q1 = data["q1"][:]     # Arm angle
    q2 = data["q2"][:]     # Pendulum angle
    u = data["u"][:]       # Motor torque

    # Plot the results
    plt.figure(figsize=(10, 4))
    plt.plot(t, q1, label="Arm Angle (q1)")
    plt.plot(t, q2, label="Pendulum Angle (q2)")
    plt.title("Pendulum Position over Time")
    plt.xlabel("Time (s)")
    plt.ylabel("Angle (radians)")
    plt.legend()
    plt.grid()
    plt.show()
```
